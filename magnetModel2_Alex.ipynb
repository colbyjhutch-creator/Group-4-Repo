```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

Temp = 0.1
size = 15

latt = np.random.choice([-0.5,0.5],size=(size,size))

#print("origin lattice is \n", latt)

#plt.imshow(latt)

def totalEnergy(latt):
    size=latt.shape[0]
    total = 0
    for i in range(size):
        for j in range(size):
            if j < size -1:
                total += latt[i][j] * latt[i][j+1]
            if i < size -1:
                total += latt[i][j] * latt[i+1][j]
    return -total

#print("total energy for original is",totalEnergy(latt))
    
def changeLatt(latt,Temp):
    latt2 = np.copy(latt)
    energy1= totalEnergy(latt)
    randomIndex1 = np.random.randint(0,latt.shape[0])
    randomIndex2 = np.random.randint(0,latt.shape[1])
    latt2[randomIndex1,randomIndex2] = -latt2[randomIndex1, randomIndex2]
   # print("new lattice is \n",latt2)
    energy2=totalEnergy(latt2)
    #print("energy1",energy1,"energy2",energy2)
    if energy2 < energy1:
        return latt2
    else:
        prob = np.exp(-(energy2-energy1)/Temp)
        #print("random number",a,"probability",prob)
        if prob > np.random.random():
            return latt2
        else:
            return latt
        
plt.rcParams["animation.html"] = "jshtml"
plt.rcParams["figure.dpi"] = 60
plt.ioff()
fig,ax = plt.subplots()

def animate(frame):
    global latt
    for i in range(20):
        latt = changeLatt(latt,Temp)
    plt.cla()
    plt.imshow(latt,cmap ="gray")
    plt.xticks([])
    plt.yticks([])
    

FuncAnimation(fig, animate, frames=200)
```
